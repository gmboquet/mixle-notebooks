# Structured science context

A model context should receive typed scientific objects and unresolved questions, not a flattened prompt. This compact example creates a mesh reference, a mathematical problem, and an explicit knowledge gap.

```python
bundle = {
    'id': 'heated-plate-context',
    'project_id': 'frontier-science-demo',
    'task': 'infer conductivity and solve the plate',
    'target_kind': 'model',
    'revision': 1,
    'items': [
        {'id': 'mesh', 'kind': 'mesh', 'schema_uri': 'mixle://schema/mesh/1', 'modality': 'mesh',
         'content_hash': '0' * 64, 'artifact_ref': 'artifact://mesh/plate-v1',
         'payload': {'topology_dimension': 2, 'geometry_dimension': 2, 'node_count': 4,
                     'cell_count': 2, 'cell_type': 'triangle', 'regions': []}},
        {'id': 'problem', 'kind': 'artifact', 'schema_uri': 'mixle://discrete/math-problem/v1', 'modality': 'graph',
         'content_hash': '1' * 64, 'payload': {'id': 'heated-plate'}},
    ],
    'gaps': [{'id': 'conductivity', 'question': 'What evidence identifies conductivity?',
              'required_schema': {'type': 'object'}, 'status': 'open'}],
    'required_capability_ids': ['mesh.consume', 'pde.solve.weak-form'],
    'handoff_policy': {'required_item_ids': ['mesh', 'problem'], 'required_gap_ids': ['conductivity'],
                       'required_modalities': ['mesh', 'graph'], 'max_item_loss_fraction': 0.0},
}
assert {item['modality'] for item in bundle['items']} == {'mesh', 'graph'}
assert bundle['gaps'][0]['status'] == 'open'
```

A receiving model must advertise both required capabilities. It returns a delta against bundle revision 1; it must not silently replace the mesh with prose or claim the conductivity gap is resolved without evidence.
